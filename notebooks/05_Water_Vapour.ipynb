# 05 Water vapour density from E-band CMLs

`scripts/wvd_validation.py` runs the unchanged WVD scripts of `original/code/wvd/` on the three WVD supporting files
of the deposit: the validation against the CHMI stations (Table S13 of the Supplement) and Fig. 8. Set the paths of
the three supporting files in the `extras` block of the configuration.

```python
from pathlib import Path
import json, subprocess, sys

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from essd_cz.config import load_config

CONFIG = ROOT / 'config' / 'user_paths.json'   # copy of config/user_paths.example.json with your paths
cfg = load_config(CONFIG)
paths, OUT = cfg['datasets'], cfg['output_dir']


def run(script, *args):
    """Run a script of the repository with the configuration of this notebook."""
    cmd = [sys.executable, str(ROOT / script), '--config', str(CONFIG), *map(str, args)]
    print(' '.join(cmd), flush=True)
    subprocess.run(cmd, check=True)
```

```python
run('scripts/wvd_validation.py')
```
